````python
import os, glob, time, sqlite3, re, pandas as pd
from openai import OpenAI

# ========= CONFIG =========
ROOT = "/content/drive/MyDrive/Dissertition/spider_2_unzipped/spider 2"
DEV_PATH = os.path.join(ROOT, "dev.json")
TABLES_PATH = os.path.join(ROOT, "tables.json")
DB_ROOT = os.path.join(ROOT, "database")
API_KEY_PATH = "/content/drive/MyDrive/Dissertition/spider_2_unzipped/openaikey.txt"

MODEL_NAME = "gpt-4o"
N_SAMPLES = None     # None = full dataset, or set an integer
MAX_REPAIRS = 3        # Increased from 2 to match code 2
SLEEP = 0.2
SHOW_ROWS = 5         # rows to print from result

# ========= INIT =========
with open(API_KEY_PATH, "r") as f:
    api_key = f.read().strip()
client = OpenAI(api_key=api_key)

df_dev = pd.read_json(DEV_PATH)
if N_SAMPLES:
    df_dev = df_dev.head(N_SAMPLES)
df_table = pd.read_json(TABLES_PATH)

# ========= Build schema {db_id: {table: [cols...]}} =========
schema_dict = {}
for _, row in df_table.iterrows():
    db_id = row["db_id"]
    tables = row["table_names_original"]
    cols = row["column_names_original"]
    schema = {t: [] for t in tables}
    for t_idx, col_name in cols:
        if t_idx >= 0:
            schema[tables[t_idx]].append(col_name)
    schema_dict[db_id] = schema

def schema_text_for(db_id: str) -> str:
    sch = schema_dict.get(db_id, {})
    return "\n".join([f"- {t}: {', '.join(cols)}" for t, cols in sch.items()])

# ========= DB helpers =========
def db_path_for(db_id: str) -> str:
    paths = [
        os.path.join(DB_ROOT, db_id, f"{db_id}.sqlite"),
        os.path.join(DB_ROOT, f"{db_id}.sqlite")
    ]
    for p in paths:
        if os.path.isfile(p):
            return p
    matches = glob.glob(os.path.join(DB_ROOT, "**", f"{db_id}.sqlite"), recursive=True)
    if matches:
        return sorted(matches)[0]   # stable order
    raise FileNotFoundError(f"Cannot find SQLite for db_id={db_id}")

def is_select_sql(sql: str) -> bool:
    return (sql or "").strip().lower().startswith(("select","with"))

def run_sql(db_id: str, sql: str):
    try:
        if not sql or not isinstance(sql, str):
            return "SQL ERROR: empty SQL"
        if not is_select_sql(sql):
            return "SQL ERROR: only SELECT/CTE queries allowed"
        conn = sqlite3.connect(db_path_for(db_id))
        conn.execute("PRAGMA query_only = ON;")
        df = pd.read_sql_query(sql, conn)
        conn.close()
        return df
    except Exception as e:
        return f"SQL ERROR: {e}"

# ========= LLM rules =========
STRICT_RULES = """You are a Text-to-SQL system for SQLite.

STRICT RULES:
1) Output ONLY ONE SQLite SELECT query.
2) Use ONLY the provided tables/columns. NEVER invent names.
3) Use explicit column names (no SELECT *).
4) NEVER use AS aliases - keep original column names.
5) For aggregates use COUNT(*), AVG(col), MIN(col), MAX(col).
6) Match golden SQL style exactly.
7) Return raw SQL only (no explanations, no code fences).
"""

SELECT_WITH_RE = re.compile(r'(?is)\b(select|with)\b.*')

def extract_sql(text: str) -> str:
    if not isinstance(text, str): return ""
    txt = text.strip().replace("```sql","").replace("```","").strip()
    m = SELECT_WITH_RE.search(txt)
    if not m: return txt
    sql = m.group(0).strip()
    if ";" in sql:
        sql = sql.split(";",1)[0] + ";"
    return sql

def llm_generate_sql(question: str, db_id: str) -> str:
    prompt = f"""{STRICT_RULES}

Schema:
{schema_text_for(db_id)}

Question:
{question}
"""
    r = client.chat.completions.create(
        model=MODEL_NAME,
        messages=[{"role":"user","content":prompt}],
        temperature=0
    )
    return extract_sql(r.choices[0].message.content)

# ========= ENHANCED REPAIR FUNCTIONS FROM CODE 2 =========
def analyze_question_intent(question: str, db_id: str) -> str:
    """Generate intelligent repair hints based on question analysis"""
    question_lower = question.lower()
    hints = []

    # Count-related patterns
    if any(word in question_lower for word in ['how many', 'count', 'number of', 'total']):
        hints.append("- This question asks for COUNT - use COUNT(*) or COUNT(column)")

    # Average patterns
    if any(word in question_lower for word in ['average', 'mean', 'avg']):
        hints.append("- This question asks for AVERAGE - use AVG(column)")

    # Maximum patterns
    if any(word in question_lower for word in ['highest', 'maximum', 'max', 'largest', 'most']):
        hints.append("- This question asks for MAXIMUM - use MAX(column)")

    # Minimum patterns
    if any(word in question_lower for word in ['lowest', 'minimum', 'min', 'smallest', 'least']):
        hints.append("- This question asks for MINIMUM - use MIN(column)")

    # Group by patterns
    if any(word in question_lower for word in ['each', 'every', 'per', 'by', 'for each']):
        hints.append("- This question asks for grouping - use GROUP BY")

    # Order by patterns
    if any(word in question_lower for word in ['order', 'sort', 'arrange', 'top', 'bottom']):
        hints.append("- This question asks for ordering - use ORDER BY")

    # Join patterns
    if any(word in question_lower for word in ['and', 'with', 'having', 'both']):
        hints.append("- This question might require JOINs between tables")

    return "\n".join(hints) if hints else "- Analyze the question structure carefully"

def generate_error_message(gold_result, pred_result):
    """Generate detailed error message for repair system"""
    if isinstance(pred_result, str):
        return pred_result

    error_parts = []

    # Shape comparison
    if gold_result.shape != pred_result.shape:
        error_parts.append(f"Shape mismatch: expected {gold_result.shape}, got {pred_result.shape}")

    # Column comparison
    gold_cols = [str(c).lower() for c in gold_result.columns]
    pred_cols = [str(c).lower() for c in pred_result.columns]
    if gold_cols != pred_cols:
        error_parts.append(f"Column mismatch: expected {list(gold_result.columns)}, got {list(pred_result.columns)}")

    # Value comparison
    if gold_result.shape == pred_result.shape and gold_cols == pred_cols:
        try:
            if gold_result.values.tolist() != pred_result.values.tolist():
                error_parts.append("Values don't match expected results")
        except:
            if not gold_result.equals(pred_result):
                error_parts.append("Values don't match expected results")

    return "RESULT ERROR: " + ("; ".join(error_parts) if error_parts else "Unknown difference")

def llm_repair_sql(question: str, db_id: str, bad_sql: str, error_msg: str, gold_result=None, pred_result=None) -> str:
    """Enhanced repair function with intelligent analysis"""
    question_hints = analyze_question_intent(question, db_id)

    prompt = f"""{STRICT_RULES}

Schema:
{schema_text_for(db_id)}

Question:
{question}

REPAIR HINTS:
{question_hints}
- Match column names EXACTLY as they appear in the schema
- Use table aliases consistently (T1, T2, etc.) or full table names
- For JOINs, ensure correct column names from both tables
- Check for typos in column/table names
- Ensure proper SQL syntax and logic

Previous SQL caused error:
{error_msg}

Previous SQL:
{bad_sql}
"""

    # Add result comparison if available
    if gold_result is not None and pred_result is not None and not isinstance(pred_result, str):
        prompt += f"""
Expected result structure:
- Rows: {gold_result.shape[0]}, Columns: {gold_result.shape[1]}
- Expected columns: {list(gold_result.columns)}

Actual result structure:
- Rows: {pred_result.shape[0]}, Columns: {pred_result.shape[1]}
- Actual columns: {list(pred_result.columns)}

Sample expected data (first few rows):
{gold_result.head(2).to_string()}
"""

    prompt += "\nReturn corrected SQL that fixes these issues."

    r = client.chat.completions.create(
        model=MODEL_NAME,
        messages=[{"role":"user","content":prompt}],
        temperature=0
    )
    return extract_sql(r.choices[0].message.content)

def predict_with_repairs(question: str, db_id: str, gold_result=None):
    """Enhanced prediction function with intelligent repair mechanism"""
    # Initial SQL generation
    sql = llm_generate_sql(question, db_id)
    out = run_sql(db_id, sql)
    attempts = 0
    repair_reasons = []
    sql_history = [sql]

    while attempts < MAX_REPAIRS:
        need_repair = False
        error_msg = None

        # Check for SQL execution errors
        if isinstance(out, str) and out.startswith("SQL ERROR"):
            need_repair = True
            error_msg = out
            repair_reasons.append(f"SQL Error")

        # Check for wrong results (if gold result is provided)
        elif gold_result is not None and not isinstance(out, str):
            exact, _, _, _ = compare(gold_result, out)
            if not exact:
                need_repair = True
                error_msg = generate_error_message(gold_result, out)
                repair_reasons.append(f"Wrong Result")

        if not need_repair:
            break

        # Attempt repair
        sql = llm_repair_sql(question, db_id, sql, error_msg, gold_result, out)
        sql_history.append(sql)
        out = run_sql(db_id, sql)
        attempts += 1
        time.sleep(SLEEP)

    return sql, out, attempts, repair_reasons, sql_history

# ========= Enhanced Comparison Function =========
def compare(gold, pred):
    """Compare gold and predicted results for exact match"""
    if isinstance(gold, str) or isinstance(pred, str):
        return False, False, False, False

    g, p = gold.reset_index(drop=True), pred.reset_index(drop=True)
    shape_ok = g.shape == p.shape

    # Compare column names (case insensitive)
    gold_cols_lower = [str(c).lower() for c in g.columns]
    pred_cols_lower = [str(c).lower() for c in p.columns]
    cols_ok = shape_ok and gold_cols_lower == pred_cols_lower

    values_match = False
    if shape_ok:
        try:
            # Convert to lists for comparison to handle different data types
            values_match = g.values.tolist() == p.values.tolist()
        except:
            # Fallback to pandas equals
            values_match = g.equals(p)

    exact = values_match and cols_ok
    return exact, cols_ok, shape_ok, values_match

tick = lambda b: "✅" if b else "❌"

def show_table(df, label):
    if isinstance(df, str):
        print(f"{label}:\n{df}")
    else:
        print(f"{label}:")
        print(df.head(SHOW_ROWS).to_string(index=False))

# ========= MAIN LOOP =========
print(f"🚀 Testing {len(df_dev)} samples with enhanced repair system...")
print("="*70)

orig_exact = imp_exact = 0
orig_mistakes, imp_mistakes = [], []
orig_repair_stats = {"successful": [], "failed": []}
imp_repair_stats = {"successful": [], "failed": []}

for i, row in df_dev.iterrows():
    db_id = row["db_id"]
    q_orig = row["question"]
    q_imp = row.get("question_improved", q_orig)
    gold_sql = row["query"]

    print(f"\n🧪 Row {i+1}/{len(df_dev)} - Database: {db_id}")

    # Golden reference
    print("\n📌 Golden SQL:")
    print(gold_sql)
    gold_out = run_sql(db_id, gold_sql)
    show_table(gold_out, "📌 Golden Table")

    # Original Q with enhanced repair system
    print(f"\n❓ Original Q: {q_orig}")
    pred_sql_o, pred_out_o, repair_attempts_o, repair_reasons_o, sql_history_o = predict_with_repairs(q_orig, db_id, gold_out)
    print("🔍 Pred SQL:", pred_sql_o)
    show_table(pred_out_o, "🔍 Pred Table")
    e, c, s, v = compare(gold_out, pred_out_o)
    print(f"🔍 Result → Exact={tick(e)} Cols={tick(c)} Shape={tick(s)} Values={tick(v)}")

    if repair_attempts_o > 0:
        print(f"🔧 Repairs attempted: {repair_attempts_o} ({', '.join(repair_reasons_o)})")
        if e:
            orig_repair_stats["successful"].append(i+1)
            print(f"✅ Successful repair!")
        else:
            orig_repair_stats["failed"].append(i+1)
            print(f"❌ Failed repair")

    if not e:
        orig_mistakes.append(i+1)
    orig_exact += e

    # Improved Q with enhanced repair system
    print(f"\n✨ Improved Q: {q_imp}")
    pred_sql_i, pred_out_i, repair_attempts_i, repair_reasons_i, sql_history_i = predict_with_repairs(q_imp, db_id, gold_out)
    print("✨ Pred SQL:", pred_sql_i)
    show_table(pred_out_i, "✨ Pred Table")
    e, c, s, v = compare(gold_out, pred_out_i)
    print(f"✨ Result → Exact={tick(e)} Cols={tick(c)} Shape={tick(s)} Values={tick(v)}")

    if repair_attempts_i > 0:
        print(f"🔧 Repairs attempted: {repair_attempts_i} ({', '.join(repair_reasons_i)})")
        if e:
            imp_repair_stats["successful"].append(i+1)
            print(f"✅ Successful repair!")
        else:
            imp_repair_stats["failed"].append(i+1)
            print(f"❌ Failed repair")

    if not e:
        imp_mistakes.append(i+1)
    imp_exact += e

    print("="*80)

# ========= ENHANCED FINAL RESULTS =========
N = len(df_dev)
print("\n📊 COMPREHENSIVE FINAL RESULTS")
print("="*80)

print(f"\n🎯 ACCURACY RESULTS:")
print(f"Original Questions Exact Match: {orig_exact}/{N} = {orig_exact/N*100:.1f}%")
print(f"Improved Questions Exact Match: {imp_exact}/{N} = {imp_exact/N*100:.1f}%")
print(f"Improvement: +{imp_exact - orig_exact} questions (+{(imp_exact - orig_exact)/N*100:.1f}% points)")

print(f"\n🔧 REPAIR STATISTICS:")
orig_total_repairs = len(orig_repair_stats["successful"]) + len(orig_repair_stats["failed"])
imp_total_repairs = len(imp_repair_stats["successful"]) + len(imp_repair_stats["failed"])

print(f"Original Questions:")
print(f"  • Total repair attempts: {orig_total_repairs}")
print(f"  • Successful repairs: {len(orig_repair_stats['successful'])}")
print(f"  • Failed repairs: {len(orig_repair_stats['failed'])}")
if orig_total_repairs > 0:
    print(f"  • Repair success rate: {len(orig_repair_stats['successful'])/orig_total_repairs*100:.1f}%")

print(f"Improved Questions:")
print(f"  • Total repair attempts: {imp_total_repairs}")
print(f"  • Successful repairs: {len(imp_repair_stats['successful'])}")
print(f"  • Failed repairs: {len(imp_repair_stats['failed'])}")
if imp_total_repairs > 0:
    print(f"  • Repair success rate: {len(imp_repair_stats['successful'])/imp_total_repairs*100:.1f}%")

print(f"\n⚠️ REMAINING MISTAKES:")
print("⚠️ ORIGINAL mistakes at rows:", orig_mistakes if orig_mistakes else "None")
print("⚠️ IMPROVED mistakes at rows:", imp_mistakes if imp_mistakes else "None")

print(f"\n✅ SUCCESSFUL REPAIRS:")
if orig_repair_stats["successful"]:
    print(f"Original questions repaired at rows: {orig_repair_stats['successful']}")
if imp_repair_stats["successful"]:
    print(f"Improved questions repaired at rows: {imp_repair_stats['successful']}")

print("\n🎉 Enhanced evaluation complete!")
````

````python
import os, glob, time, sqlite3, re, pandas as pd
from openai import OpenAI

# ========= CONFIG =========
ROOT = "/content/drive/MyDrive/Dissertition/spider_2_unzipped/spider 2"
DEV_PATH = os.path.join(ROOT, "dev.json")
TABLES_PATH = os.path.join(ROOT, "tables.json")
DB_ROOT = os.path.join(ROOT, "database")
API_KEY_PATH = "/content/drive/MyDrive/Dissertition/spider_2_unzipped/openaikey.txt"

MODEL_NAME = "gpt-4o-mini"
N_SAMPLES = None    # None = full dataset, or set an integer
MAX_REPAIRS = 3        # Increased from 2 to match code 2
SLEEP = 0.2
SHOW_ROWS = 5         # rows to print from result

# ========= INIT =========
with open(API_KEY_PATH, "r") as f:
    api_key = f.read().strip()
client = OpenAI(api_key=api_key)

df_dev = pd.read_json(DEV_PATH)
if N_SAMPLES:
    df_dev = df_dev.head(N_SAMPLES)
df_table = pd.read_json(TABLES_PATH)

# ========= Build schema {db_id: {table: [cols...]}} =========
schema_dict = {}
for _, row in df_table.iterrows():
    db_id = row["db_id"]
    tables = row["table_names_original"]
    cols = row["column_names_original"]
    schema = {t: [] for t in tables}
    for t_idx, col_name in cols:
        if t_idx >= 0:
            schema[tables[t_idx]].append(col_name)
    schema_dict[db_id] = schema

def schema_text_for(db_id: str) -> str:
    sch = schema_dict.get(db_id, {})
    return "\n".join([f"- {t}: {', '.join(cols)}" for t, cols in sch.items()])

# ========= DB helpers =========
def db_path_for(db_id: str) -> str:
    paths = [
        os.path.join(DB_ROOT, db_id, f"{db_id}.sqlite"),
        os.path.join(DB_ROOT, f"{db_id}.sqlite")
    ]
    for p in paths:
        if os.path.isfile(p):
            return p
    matches = glob.glob(os.path.join(DB_ROOT, "**", f"{db_id}.sqlite"), recursive=True)
    if matches:
        return sorted(matches)[0]   # stable order
    raise FileNotFoundError(f"Cannot find SQLite for db_id={db_id}")

def is_select_sql(sql: str) -> bool:
    return (sql or "").strip().lower().startswith(("select","with"))

def run_sql(db_id: str, sql: str):
    try:
        if not sql or not isinstance(sql, str):
            return "SQL ERROR: empty SQL"
        if not is_select_sql(sql):
            return "SQL ERROR: only SELECT/CTE queries allowed"
        conn = sqlite3.connect(db_path_for(db_id))
        conn.execute("PRAGMA query_only = ON;")
        df = pd.read_sql_query(sql, conn)
        conn.close()
        return df
    except Exception as e:
        return f"SQL ERROR: {e}"

# ========= LLM rules =========
STRICT_RULES = """You are a Text-to-SQL system for SQLite.

STRICT RULES:
1) Output ONLY ONE SQLite SELECT query.
2) Use ONLY the provided tables/columns. NEVER invent names.
3) Use explicit column names (no SELECT *).
4) NEVER use AS aliases - keep original column names.
5) For aggregates use COUNT(*), AVG(col), MIN(col), MAX(col).
6) Match golden SQL style exactly.
7) Return raw SQL only (no explanations, no code fences).
"""

SELECT_WITH_RE = re.compile(r'(?is)\b(select|with)\b.*')

def extract_sql(text: str) -> str:
    if not isinstance(text, str): return ""
    txt = text.strip().replace("```sql","").replace("```","").strip()
    m = SELECT_WITH_RE.search(txt)
    if not m: return txt
    sql = m.group(0).strip()
    if ";" in sql:
        sql = sql.split(";",1)[0] + ";"
    return sql

def llm_generate_sql(question: str, db_id: str) -> str:
    prompt = f"""{STRICT_RULES}

Schema:
{schema_text_for(db_id)}

Question:
{question}
"""
    r = client.chat.completions.create(
        model=MODEL_NAME,
        messages=[{"role":"user","content":prompt}],
        temperature=0
    )
    return extract_sql(r.choices[0].message.content)

# ========= ENHANCED REPAIR FUNCTIONS FROM CODE 2 =========
def analyze_question_intent(question: str, db_id: str) -> str:
    """Generate intelligent repair hints based on question analysis"""
    question_lower = question.lower()
    hints = []

    # Count-related patterns
    if any(word in question_lower for word in ['how many', 'count', 'number of', 'total']):
        hints.append("- This question asks for COUNT - use COUNT(*) or COUNT(column)")

    # Average patterns
    if any(word in question_lower for word in ['average', 'mean', 'avg']):
        hints.append("- This question asks for AVERAGE - use AVG(column)")

    # Maximum patterns
    if any(word in question_lower for word in ['highest', 'maximum', 'max', 'largest', 'most']):
        hints.append("- This question asks for MAXIMUM - use MAX(column)")

    # Minimum patterns
    if any(word in question_lower for word in ['lowest', 'minimum', 'min', 'smallest', 'least']):
        hints.append("- This question asks for MINIMUM - use MIN(column)")

    # Group by patterns
    if any(word in question_lower for word in ['each', 'every', 'per', 'by', 'for each']):
        hints.append("- This question asks for grouping - use GROUP BY")

    # Order by patterns
    if any(word in question_lower for word in ['order', 'sort', 'arrange', 'top', 'bottom']):
        hints.append("- This question asks for ordering - use ORDER BY")

    # Join patterns
    if any(word in question_lower for word in ['and', 'with', 'having', 'both']):
        hints.append("- This question might require JOINs between tables")

    return "\n".join(hints) if hints else "- Analyze the question structure carefully"

def generate_error_message(gold_result, pred_result):
    """Generate detailed error message for repair system"""
    if isinstance(pred_result, str):
        return pred_result

    error_parts = []

    # Shape comparison
    if gold_result.shape != pred_result.shape:
        error_parts.append(f"Shape mismatch: expected {gold_result.shape}, got {pred_result.shape}")

    # Column comparison
    gold_cols = [str(c).lower() for c in gold_result.columns]
    pred_cols = [str(c).lower() for c in pred_result.columns]
    if gold_cols != pred_cols:
        error_parts.append(f"Column mismatch: expected {list(gold_result.columns)}, got {list(pred_result.columns)}")

    # Value comparison
    if gold_result.shape == pred_result.shape and gold_cols == pred_cols:
        try:
            if gold_result.values.tolist() != pred_result.values.tolist():
                error_parts.append("Values don't match expected results")
        except:
            if not gold_result.equals(pred_result):
                error_parts.append("Values don't match expected results")

    return "RESULT ERROR: " + ("; ".join(error_parts) if error_parts else "Unknown difference")

def llm_repair_sql(question: str, db_id: str, bad_sql: str, error_msg: str, gold_result=None, pred_result=None) -> str:
    """Enhanced repair function with intelligent analysis"""
    question_hints = analyze_question_intent(question, db_id)

    prompt = f"""{STRICT_RULES}

Schema:
{schema_text_for(db_id)}

Question:
{question}

REPAIR HINTS:
{question_hints}
- Match column names EXACTLY as they appear in the schema
- Use table aliases consistently (T1, T2, etc.) or full table names
- For JOINs, ensure correct column names from both tables
- Check for typos in column/table names
- Ensure proper SQL syntax and logic

Previous SQL caused error:
{error_msg}

Previous SQL:
{bad_sql}
"""

    # Add result comparison if available
    if gold_result is not None and pred_result is not None and not isinstance(pred_result, str):
        prompt += f"""
Expected result structure:
- Rows: {gold_result.shape[0]}, Columns: {gold_result.shape[1]}
- Expected columns: {list(gold_result.columns)}

Actual result structure:
- Rows: {pred_result.shape[0]}, Columns: {pred_result.shape[1]}
- Actual columns: {list(pred_result.columns)}

Sample expected data (first few rows):
{gold_result.head(2).to_string()}
"""

    prompt += "\nReturn corrected SQL that fixes these issues."

    r = client.chat.completions.create(
        model=MODEL_NAME,
        messages=[{"role":"user","content":prompt}],
        temperature=0
    )
    return extract_sql(r.choices[0].message.content)

def predict_with_repairs(question: str, db_id: str, gold_result=None):
    """Enhanced prediction function with intelligent repair mechanism"""
    # Initial SQL generation
    sql = llm_generate_sql(question, db_id)
    out = run_sql(db_id, sql)
    attempts = 0
    repair_reasons = []
    sql_history = [sql]

    while attempts < MAX_REPAIRS:
        need_repair = False
        error_msg = None

        # Check for SQL execution errors
        if isinstance(out, str) and out.startswith("SQL ERROR"):
            need_repair = True
            error_msg = out
            repair_reasons.append(f"SQL Error")

        # Check for wrong results (if gold result is provided)
        elif gold_result is not None and not isinstance(out, str):
            exact, _, _, _ = compare(gold_result, out)
            if not exact:
                need_repair = True
                error_msg = generate_error_message(gold_result, out)
                repair_reasons.append(f"Wrong Result")

        if not need_repair:
            break

        # Attempt repair
        sql = llm_repair_sql(question, db_id, sql, error_msg, gold_result, out)
        sql_history.append(sql)
        out = run_sql(db_id, sql)
        attempts += 1
        time.sleep(SLEEP)

    return sql, out, attempts, repair_reasons, sql_history

# ========= Enhanced Comparison Function =========
def compare(gold, pred):
    """Compare gold and predicted results for exact match"""
    if isinstance(gold, str) or isinstance(pred, str):
        return False, False, False, False

    g, p = gold.reset_index(drop=True), pred.reset_index(drop=True)
    shape_ok = g.shape == p.shape

    # Compare column names (case insensitive)
    gold_cols_lower = [str(c).lower() for c in g.columns]
    pred_cols_lower = [str(c).lower() for c in p.columns]
    cols_ok = shape_ok and gold_cols_lower == pred_cols_lower

    values_match = False
    if shape_ok:
        try:
            # Convert to lists for comparison to handle different data types
            values_match = g.values.tolist() == p.values.tolist()
        except:
            # Fallback to pandas equals
            values_match = g.equals(p)

    exact = values_match and cols_ok
    return exact, cols_ok, shape_ok, values_match

tick = lambda b: "✅" if b else "❌"

def show_table(df, label):
    if isinstance(df, str):
        print(f"{label}:\n{df}")
    else:
        print(f"{label}:")
        print(df.head(SHOW_ROWS).to_string(index=False))

# ========= MAIN LOOP =========
print(f"🚀 Testing {len(df_dev)} samples with enhanced repair system...")
print("="*70)

orig_exact = imp_exact = 0
orig_mistakes, imp_mistakes = [], []
orig_repair_stats = {"successful": [], "failed": []}
imp_repair_stats = {"successful": [], "failed": []}

# Variables for future analysis - GPT-4o Mini specific
original_mistakes_4o_mini = []
improved_mistakes_4o_mini = []

for i, row in df_dev.iterrows():
    db_id = row["db_id"]
    q_orig = row["question"]
    q_imp = row.get("question_improved", q_orig)
    gold_sql = row["query"]

    print(f"\n🧪 Row {i+1}/{len(df_dev)} - Database: {db_id}")

    # Golden reference
    print("\n📌 Golden SQL:")
    print(gold_sql)
    gold_out = run_sql(db_id, gold_sql)
    show_table(gold_out, "📌 Golden Table")

    # Original Q with enhanced repair system
    print(f"\n❓ Original Q: {q_orig}")
    pred_sql_o, pred_out_o, repair_attempts_o, repair_reasons_o, sql_history_o = predict_with_repairs(q_orig, db_id, gold_out)
    print("🔍 Pred SQL:", pred_sql_o)
    show_table(pred_out_o, "🔍 Pred Table")
    e, c, s, v = compare(gold_out, pred_out_o)
    print(f"🔍 Result → Exact={tick(e)} Cols={tick(c)} Shape={tick(s)} Values={tick(v)}")

    if repair_attempts_o > 0:
        print(f"🔧 Repairs attempted: {repair_attempts_o} ({', '.join(repair_reasons_o)})")
        if e:
            orig_repair_stats["successful"].append(i+1)
            print(f"✅ Successful repair!")
        else:
            orig_repair_stats["failed"].append(i+1)
            print(f"❌ Failed repair")

    if not e:
        orig_mistakes.append(i+1)
        original_mistakes_4o_mini.append(i+1)  # Save for future analysis
    orig_exact += e

    # Improved Q with enhanced repair system
    print(f"\n✨ Improved Q: {q_imp}")
    pred_sql_i, pred_out_i, repair_attempts_i, repair_reasons_i, sql_history_i = predict_with_repairs(q_imp, db_id, gold_out)
    print("✨ Pred SQL:", pred_sql_i)
    show_table(pred_out_i, "✨ Pred Table")
    e, c, s, v = compare(gold_out, pred_out_i)
    print(f"✨ Result → Exact={tick(e)} Cols={tick(c)} Shape={tick(s)} Values={tick(v)}")

    if repair_attempts_i > 0:
        print(f"🔧 Repairs attempted: {repair_attempts_i} ({', '.join(repair_reasons_i)})")
        if e:
            imp_repair_stats["successful"].append(i+1)
            print(f"✅ Successful repair!")
        else:
            imp_repair_stats["failed"].append(i+1)
            print(f"❌ Failed repair")

    if not e:
        imp_mistakes.append(i+1)
        improved_mistakes_4o_mini.append(i+1)  # Save for future analysis
    imp_exact += e

    print("="*80)

# ========= ENHANCED FINAL RESULTS =========
N = len(df_dev)
print("\n📊 COMPREHENSIVE FINAL RESULTS")
print("="*80)

print(f"\n🎯 ACCURACY RESULTS:")
print(f"Original Questions Exact Match: {orig_exact}/{N} = {orig_exact/N*100:.1f}%")
print(f"Improved Questions Exact Match: {imp_exact}/{N} = {imp_exact/N*100:.1f}%")
print(f"Improvement: +{imp_exact - orig_exact} questions (+{(imp_exact - orig_exact)/N*100:.1f}% points)")

print(f"\n🔧 REPAIR STATISTICS:")
orig_total_repairs = len(orig_repair_stats["successful"]) + len(orig_repair_stats["failed"])
imp_total_repairs = len(imp_repair_stats["successful"]) + len(imp_repair_stats["failed"])

print(f"Original Questions:")
print(f"  • Total repair attempts: {orig_total_repairs}")
print(f"  • Successful repairs: {len(orig_repair_stats['successful'])}")
print(f"  • Failed repairs: {len(orig_repair_stats['failed'])}")
if orig_total_repairs > 0:
    print(f"  • Repair success rate: {len(orig_repair_stats['successful'])/orig_total_repairs*100:.1f}%")

print(f"Improved Questions:")
print(f"  • Total repair attempts: {imp_total_repairs}")
print(f"  • Successful repairs: {len(imp_repair_stats['successful'])}")
print(f"  • Failed repairs: {len(imp_repair_stats['failed'])}")
if imp_total_repairs > 0:
    print(f"  • Repair success rate: {len(imp_repair_stats['successful'])/imp_total_repairs*100:.1f}%")

print(f"\n⚠️ REMAINING MISTAKES:")
print("⚠️ ORIGINAL mistakes at rows:", orig_mistakes if orig_mistakes else "None")
print("⚠️ IMPROVED mistakes at rows:", imp_mistakes if imp_mistakes else "None")

print(f"\n✅ SUCCESSFUL REPAIRS:")
if orig_repair_stats["successful"]:
    print(f"Original questions repaired at rows: {orig_repair_stats['successful']}")
if imp_repair_stats["successful"]:
    print(f"Improved questions repaired at rows: {imp_repair_stats['successful']}")

# Save mistake rows for future analysis
print(f"\n💾 SAVED FOR FUTURE ANALYSIS:")
print(f"original_mistakes_4o_mini = {original_mistakes_4o_mini}")
print(f"improved_mistakes_4o_mini = {improved_mistakes_4o_mini}")

print("\n🎉 Enhanced evaluation complete!")
````

````python
import os, glob, time, sqlite3, re, pandas as pd
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# ========= ULTRA-FAST CONFIG =========
ROOT = "/content/drive/MyDrive/Dissertition/spider_2_unzipped/spider 2"
DEV_PATH = os.path.join(ROOT, "dev.json")
TABLES_PATH = os.path.join(ROOT, "tables.json")
DB_ROOT = os.path.join(ROOT, "database")

MODEL_NAME = "codellama/CodeLlama-7b-Instruct-hf"
N_SAMPLES = None
MAX_REPAIRS = 1  # Reduced repairs for speed
SHOW_ROWS = 3  # Fewer rows to display

# ========= LIGHTNING-FAST MODEL SETUP =========
print("Loading model with maximum speed optimizations...")

# Ultra-fast tokenizer
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, use_fast=True, padding_side="left")
tokenizer.pad_token = tokenizer.eos_token

# Load model with extreme speed settings
try:
    model = AutoModelForCausalLM.from_pretrained(
        MODEL_NAME,
        torch_dtype=torch.float16,
        device_map="auto",
        low_cpu_mem_usage=True,
        trust_remote_code=True,
        # attn_implementation="flash_attention_2"  # Uncomment if available
    )
except Exception as e:
    print(f"Loading with flash attention failed: {e}")
    print("Loading with standard settings...")
    model = AutoModelForCausalLM.from_pretrained(
        MODEL_NAME,
        torch_dtype=torch.float16,
        device_map="auto",
        low_cpu_mem_usage=True,
        trust_remote_code=True
    )

model.eval()
print(f"Model loaded successfully on device: {model.device}")

# ========= INIT =========
df_dev = pd.read_json(DEV_PATH)
if N_SAMPLES:
    df_dev = df_dev.head(N_SAMPLES)
df_table = pd.read_json(TABLES_PATH)

# ========= Build schema {db_id: {table: [cols...]}} =========
schema_dict = {}
for _, row in df_table.iterrows():
    db_id = row["db_id"]
    tables = row["table_names_original"]
    cols = row["column_names_original"]
    schema = {t: [] for t in tables}
    for t_idx, col_name in cols:
        if t_idx >= 0:
            schema[tables[t_idx]].append(col_name)
    schema_dict[db_id] = schema

def schema_text_for(db_id: str) -> str:
    sch = schema_dict.get(db_id, {})
    return "\n".join([f"- {t}: {', '.join(cols)}" for t, cols in sch.items()])

# ========= DB helpers =========
def db_path_for(db_id: str) -> str:
    paths = [
        os.path.join(DB_ROOT, db_id, f"{db_id}.sqlite"),
        os.path.join(DB_ROOT, f"{db_id}.sqlite")
    ]
    for p in paths:
        if os.path.isfile(p):
            return p
    matches = glob.glob(os.path.join(DB_ROOT, "**", f"{db_id}.sqlite"), recursive=True)
    if matches:
        return sorted(matches)[0]
    raise FileNotFoundError(f"Cannot find SQLite for db_id={db_id}")

def is_select_sql(sql: str) -> bool:
    return (sql or "").strip().lower().startswith(("select","with"))

def run_sql(db_id: str, sql: str):
    try:
        if not sql or not isinstance(sql, str):
            return "SQL ERROR: empty SQL"
        if not is_select_sql(sql):
            return "SQL ERROR: only SELECT/CTE queries allowed"
        conn = sqlite3.connect(db_path_for(db_id))
        conn.execute("PRAGMA query_only = ON;")
        df = pd.read_sql_query(sql, conn)
        conn.close()
        return df
    except Exception as e:
        return f"SQL ERROR: {e}"

# ========= LLAMA GENERATION FUNCTIONS =========
def generate_with_llama(prompt: str, max_new_tokens: int = 256, temperature: float = 0.1) -> str:
    """Generate response using local Llama model"""
    try:
        # Format prompt for CodeLlama Instruct
        formatted_prompt = f"[INST] {prompt} [/INST]"

        # Tokenize input
        inputs = tokenizer(formatted_prompt, return_tensors="pt", truncation=True, max_length=2048)

        # Move to same device as model
        inputs = {k: v.to(model.device) for k, v in inputs.items()}

        # Generate with optimized settings
        with torch.no_grad():
            outputs = model.generate(
                **inputs,
                max_new_tokens=max_new_tokens,
                temperature=temperature,
                do_sample=temperature > 0,
                pad_token_id=tokenizer.eos_token_id,
                eos_token_id=tokenizer.eos_token_id,
                use_cache=True
            )

        # Decode response
        response = tokenizer.decode(outputs[0], skip_special_tokens=True)

        # Extract only the new generated part
        response = response[len(formatted_prompt):].strip()

        return response

    except Exception as e:
        print(f"Generation error: {e}")
        return ""

# ========= LLM rules =========
STRICT_RULES = """You are a Text-to-SQL system for SQLite.

STRICT RULES:
1) Output ONLY ONE SQLite SELECT query.
2) Use ONLY the provided tables/columns. NEVER invent names.
3) Use explicit column names (no SELECT *).
4) NEVER use AS aliases - keep original column names.
5) For aggregates use COUNT(*), AVG(col), MIN(col), MAX(col).
6) Return raw SQL only (no explanations, no code fences, no markdown).
7) Do not include any text before or after the SQL query.

Schema:"""

SELECT_WITH_RE = re.compile(r'(?is)\b(select|with)\b.*')

def extract_sql(text: str) -> str:
    if not isinstance(text, str):
        return ""

    # Clean the text
    txt = text.strip()

    # Remove common markdown code fences
    txt = re.sub(r'```sql\s*', '', txt)
    txt = re.sub(r'```\s*', '', txt)

    # Try to find SQL starting with SELECT or WITH
    lines = txt.split('\n')
    sql_lines = []

    for line in lines:
        line = line.strip()
        if line.lower().startswith(('select', 'with')):
            # Found start of SQL
            sql_lines = [line]
            continue
        elif sql_lines:
            # We're in SQL mode
            if line.endswith(';'):
                sql_lines.append(line)
                break
            elif any(keyword in line.lower() for keyword in ['from', 'where', 'group by', 'having', 'order by', 'limit', 'join', 'union', 'and', 'or']):
                sql_lines.append(line)
            elif line == '':
                continue  # Skip empty lines within SQL
            else:
                # Might be end of SQL
                break

    if sql_lines:
        sql = ' '.join(sql_lines)
    else:
        # Fallback to regex
        m = SELECT_WITH_RE.search(txt)
        if m:
            sql = m.group(0).strip()
        else:
            sql = txt

    # Clean up
    if ";" in sql:
        sql = sql.split(";")[0] + ";"

    return sql.strip()

def llm_generate_sql(question: str, db_id: str) -> str:
    prompt = f"""{STRICT_RULES}

{schema_text_for(db_id)}

Question: {question}

SQL:"""

    response = generate_with_llama(prompt, max_new_tokens=128, temperature=0.1)
    return extract_sql(response)

# ========= ENHANCED REPAIR FUNCTIONS =========
def analyze_question_intent(question: str, db_id: str) -> str:
    """Generate intelligent repair hints based on question analysis"""
    question_lower = question.lower()
    hints = []

    if any(word in question_lower for word in ['how many', 'count', 'number of', 'total']):
        hints.append("- This question asks for COUNT - use COUNT(*) or COUNT(column)")

    if any(word in question_lower for word in ['average', 'mean', 'avg']):
        hints.append("- This question asks for AVERAGE - use AVG(column)")

    if any(word in question_lower for word in ['highest', 'maximum', 'max', 'largest', 'most']):
        hints.append("- This question asks for MAXIMUM - use MAX(column)")

    if any(word in question_lower for word in ['lowest', 'minimum', 'min', 'smallest', 'least']):
        hints.append("- This question asks for MINIMUM - use MIN(column)")

    if any(word in question_lower for word in ['each', 'every', 'per', 'by', 'for each']):
        hints.append("- This question asks for grouping - use GROUP BY")

    if any(word in question_lower for word in ['order', 'sort', 'arrange', 'top', 'bottom']):
        hints.append("- This question asks for ordering - use ORDER BY")

    if any(word in question_lower for word in ['and', 'with', 'having', 'both']):
        hints.append("- This question might require JOINs between tables")

    return "\n".join(hints) if hints else "- Analyze the question structure carefully"

def generate_error_message(gold_result, pred_result):
    """Generate detailed error message for repair system"""
    if isinstance(pred_result, str):
        return pred_result

    error_parts = []

    if gold_result.shape != pred_result.shape:
        error_parts.append(f"Shape mismatch: expected {gold_result.shape}, got {pred_result.shape}")

    gold_cols = [str(c).lower() for c in gold_result.columns]
    pred_cols = [str(c).lower() for c in pred_result.columns]
    if gold_cols != pred_cols:
        error_parts.append(f"Column mismatch: expected {list(gold_result.columns)}, got {list(pred_result.columns)}")

    if gold_result.shape == pred_result.shape and gold_cols == pred_cols:
        try:
            if gold_result.values.tolist() != pred_result.values.tolist():
                error_parts.append("Values don't match expected results")
        except:
            if not gold_result.equals(pred_result):
                error_parts.append("Values don't match expected results")

    return "RESULT ERROR: " + ("; ".join(error_parts) if error_parts else "Unknown difference")

def llm_repair_sql(question: str, db_id: str, bad_sql: str, error_msg: str, gold_result=None, pred_result=None) -> str:
    """Enhanced repair function with intelligent analysis"""
    question_hints = analyze_question_intent(question, db_id)

    prompt = f"""{STRICT_RULES}

{schema_text_for(db_id)}

Question: {question}

REPAIR HINTS:
{question_hints}
- Match column names EXACTLY as they appear in the schema
- Check for typos in column/table names
- Ensure proper SQL syntax and logic

Previous SQL caused error: {error_msg}

Previous SQL: {bad_sql}

Corrected SQL:"""

    response = generate_with_llama(prompt, max_new_tokens=128, temperature=0.1)
    return extract_sql(response)

def predict_with_repairs(question: str, db_id: str, gold_result=None):
    """Enhanced prediction function with repair mechanism"""
    sql = llm_generate_sql(question, db_id)
    out = run_sql(db_id, sql)
    attempts = 0
    repair_reasons = []
    sql_history = [sql]

    while attempts < MAX_REPAIRS:
        need_repair = False
        error_msg = None

        if isinstance(out, str) and out.startswith("SQL ERROR"):
            need_repair = True
            error_msg = out
            repair_reasons.append("SQL Error")

        elif gold_result is not None and not isinstance(out, str):
            exact, _, _, _ = compare(gold_result, out)
            if not exact:
                need_repair = True
                error_msg = generate_error_message(gold_result, out)
                repair_reasons.append("Wrong Result")

        if not need_repair:
            break

        sql = llm_repair_sql(question, db_id, sql, error_msg, gold_result, out)
        sql_history.append(sql)
        out = run_sql(db_id, sql)
        attempts += 1

    return sql, out, attempts, repair_reasons, sql_history

# ========= Comparison Function =========
def compare(gold, pred):
    """Compare gold and predicted results for exact match"""
    if isinstance(gold, str) or isinstance(pred, str):
        return False, False, False, False

    g, p = gold.reset_index(drop=True), pred.reset_index(drop=True)
    shape_ok = g.shape == p.shape

    gold_cols_lower = [str(c).lower() for c in g.columns]
    pred_cols_lower = [str(c).lower() for c in p.columns]
    cols_ok = shape_ok and gold_cols_lower == pred_cols_lower

    values_match = False
    if shape_ok:
        try:
            values_match = g.values.tolist() == p.values.tolist()
        except:
            values_match = g.equals(p)

    exact = values_match and cols_ok
    return exact, cols_ok, shape_ok, values_match

tick = lambda b: "✅" if b else "❌"

def show_table(df, label):
    if isinstance(df, str):
        print(f"{label}:\n{df}")
    else:
        print(f"{label}:")
        print(df.head(SHOW_ROWS).to_string(index=False))

# ========= MAIN LOOP =========
print(f"🚀 Testing {len(df_dev)} samples with CodeLlama model")
print("="*70)

orig_exact = imp_exact = 0
orig_mistakes, imp_mistakes = [], []
orig_repair_stats = {"successful": [], "failed": []}
imp_repair_stats = {"successful": [], "failed": []}

for i, row in df_dev.iterrows():
    db_id = row["db_id"]
    q_orig = row["question"]
    q_imp = row.get("question_improved", q_orig)
    gold_sql = row["query"]

    print(f"\n🧪 Row {i+1}/{len(df_dev)} - Database: {db_id}")

    # Golden reference
    print("\n📌 Golden SQL:")
    print(gold_sql)
    gold_out = run_sql(db_id, gold_sql)
    show_table(gold_out, "📌 Golden Table")

    # Original Q with repair system
    print(f"\n❓ Original Q: {q_orig}")
    pred_sql_o, pred_out_o, repair_attempts_o, repair_reasons_o, sql_history_o = predict_with_repairs(q_orig, db_id, gold_out)
    print("🔍 Pred SQL:", pred_sql_o)
    show_table(pred_out_o, "🔍 Pred Table")
    e, c, s, v = compare(gold_out, pred_out_o)
    print(f"🔍 Result → Exact={tick(e)} Cols={tick(c)} Shape={tick(s)} Values={tick(v)}")

    if repair_attempts_o > 0:
        print(f"🔧 Repairs attempted: {repair_attempts_o} ({', '.join(repair_reasons_o)})")
        if e:
            orig_repair_stats["successful"].append(i+1)
            print("✅ Successful repair!")
        else:
            orig_repair_stats["failed"].append(i+1)
            print("❌ Failed repair")

    if not e:
        orig_mistakes.append(i+1)
    orig_exact += e

    # Improved Q with repair system
    print(f"\n✨ Improved Q: {q_imp}")
    pred_sql_i, pred_out_i, repair_attempts_i, repair_reasons_i, sql_history_i = predict_with_repairs(q_imp, db_id, gold_out)
    print("✨ Pred SQL:", pred_sql_i)
    show_table(pred_out_i, "✨ Pred Table")
    e, c, s, v = compare(gold_out, pred_out_i)
    print(f"✨ Result → Exact={tick(e)} Cols={tick(c)} Shape={tick(s)} Values={tick(v)}")

    if repair_attempts_i > 0:
        print(f"🔧 Repairs attempted: {repair_attempts_i} ({', '.join(repair_reasons_i)})")
        if e:
            imp_repair_stats["successful"].append(i+1)
            print("✅ Successful repair!")
        else:
            imp_repair_stats["failed"].append(i+1)
            print("❌ Failed repair")

    if not e:
        imp_mistakes.append(i+1)
    imp_exact += e

    print("="*80)

# ========= FINAL RESULTS =========
N = len(df_dev)
print(f"\n📊 FINAL RESULTS")
print("="*80)

print(f"\n🎯 ACCURACY RESULTS:")
print(f"Original Questions Exact Match: {orig_exact}/{N} = {orig_exact/N*100:.1f}%")
print(f"Improved Questions Exact Match: {imp_exact}/{N} = {imp_exact/N*100:.1f}%")
print(f"Improvement: +{imp_exact - orig_exact} questions (+{(imp_exact - orig_exact)/N*100:.1f}% points)")

print(f"\n🔧 REPAIR STATISTICS:")
orig_total_repairs = len(orig_repair_stats["successful"]) + len(orig_repair_stats["failed"])
imp_total_repairs = len(imp_repair_stats["successful"]) + len(imp_repair_stats["failed"])

print(f"Original Questions:")
print(f"  • Total repair attempts: {orig_total_repairs}")
print(f"  • Successful repairs: {len(orig_repair_stats['successful'])}")
print(f"  • Failed repairs: {len(orig_repair_stats['failed'])}")
if orig_total_repairs > 0:
    print(f"  • Repair success rate: {len(orig_repair_stats['successful'])/orig_total_repairs*100:.1f}%")

print(f"Improved Questions:")
print(f"  • Total repair attempts: {imp_total_repairs}")
print(f"  • Successful repairs: {len(imp_repair_stats['successful'])}")
print(f"  • Failed repairs: {len(imp_repair_stats['failed'])}")
if imp_total_repairs > 0:
    print(f"  • Repair success rate: {len(imp_repair_stats['successful'])/imp_total_repairs*100:.1f}%")

print(f"\n⚠️ REMAINING MISTAKES:")
print("⚠️ ORIGINAL mistakes at rows:", orig_mistakes if orig_mistakes else "None")
print("⚠️ IMPROVED mistakes at rows:", imp_mistakes if imp_mistakes else "None")

print(f"\n✅ SUCCESSFUL REPAIRS:")
if orig_repair_stats["successful"]:
    print(f"Original questions repaired at rows: {orig_repair_stats['successful']}")
if imp_repair_stats["successful"]:
    print(f"Improved questions repaired at rows: {imp_repair_stats['successful']}")

print(f"\n🎉 Evaluation complete with CodeLlama model!")

# Clean up GPU memory
if torch.cuda.is_available():
    torch.cuda.empty_cache()
````

````python
import os, glob, time, sqlite3, re, pandas as pd
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# ========= ULTRA-FAST CONFIG =========
ROOT = "/content/drive/MyDrive/Dissertition/spider_2_unzipped/spider 2"
DEV_PATH = os.path.join(ROOT, "dev.json")
TABLES_PATH = os.path.join(ROOT, "tables.json")
DB_ROOT = os.path.join(ROOT, "database")

MODEL_NAME = "Qwen/Qwen2.5-Coder-1.5B-Instruct"
N_SAMPLES = None
MAX_REPAIRS = 1  # Reduced repairs for speed
SHOW_ROWS = 3  # Fewer rows to display

# ========= LIGHTNING-FAST MODEL SETUP =========
print("Loading model with maximum speed optimizations...")

# Ultra-fast tokenizer
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, use_fast=True, padding_side="left")
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

# Load model with extreme speed settings
try:
    model = AutoModelForCausalLM.from_pretrained(
        MODEL_NAME,
        torch_dtype=torch.float16,
        device_map="auto",
        low_cpu_mem_usage=True,
        trust_remote_code=True,
        # attn_implementation="flash_attention_2"  # Uncomment if available
    )
except Exception as e:
    print(f"Loading with flash attention failed: {e}")
    print("Loading with standard settings...")
    model = AutoModelForCausalLM.from_pretrained(
        MODEL_NAME,
        torch_dtype=torch.float16,
        device_map="auto",
        low_cpu_mem_usage=True,
        trust_remote_code=True
    )

model.eval()
print(f"Model loaded successfully on device: {model.device}")

# ========= INIT =========
df_dev = pd.read_json(DEV_PATH)
if N_SAMPLES:
    df_dev = df_dev.head(N_SAMPLES)
df_table = pd.read_json(TABLES_PATH)

# ========= Build schema {db_id: {table: [cols...]}} =========
schema_dict = {}
for _, row in df_table.iterrows():
    db_id = row["db_id"]
    tables = row["table_names_original"]
    cols = row["column_names_original"]
    schema = {t: [] for t in tables}
    for t_idx, col_name in cols:
        if t_idx >= 0:
            schema[tables[t_idx]].append(col_name)
    schema_dict[db_id] = schema

def schema_text_for(db_id: str) -> str:
    sch = schema_dict.get(db_id, {})
    return "\n".join([f"- {t}: {', '.join(cols)}" for t, cols in sch.items()])

# ========= DB helpers =========
def db_path_for(db_id: str) -> str:
    paths = [
        os.path.join(DB_ROOT, db_id, f"{db_id}.sqlite"),
        os.path.join(DB_ROOT, f"{db_id}.sqlite")
    ]
    for p in paths:
        if os.path.isfile(p):
            return p
    matches = glob.glob(os.path.join(DB_ROOT, "**", f"{db_id}.sqlite"), recursive=True)
    if matches:
        return sorted(matches)[0]
    raise FileNotFoundError(f"Cannot find SQLite for db_id={db_id}")

def is_select_sql(sql: str) -> bool:
    return (sql or "").strip().lower().startswith(("select","with"))

def run_sql(db_id: str, sql: str):
    try:
        if not sql or not isinstance(sql, str):
            return "SQL ERROR: empty SQL"
        if not is_select_sql(sql):
            return "SQL ERROR: only SELECT/CTE queries allowed"
        conn = sqlite3.connect(db_path_for(db_id))
        conn.execute("PRAGMA query_only = ON;")
        df = pd.read_sql_query(sql, conn)
        conn.close()
        return df
    except Exception as e:
        return f"SQL ERROR: {e}"

# ========= QWEN GENERATION FUNCTIONS =========
def generate_with_qwen(prompt: str, max_new_tokens: int = 256, temperature: float = 0.1) -> str:
    """Generate response using local Qwen model"""
    try:
        # Tokenize input
        inputs = tokenizer(prompt, return_tensors="pt", truncation=True, max_length=2048)

        # Move to same device as model
        inputs = {k: v.to(model.device) for k, v in inputs.items()}

        # Generate with optimized settings
        with torch.no_grad():
            outputs = model.generate(
                **inputs,
                max_new_tokens=max_new_tokens,
                temperature=temperature,
                do_sample=temperature > 0,
                pad_token_id=tokenizer.eos_token_id,
                eos_token_id=tokenizer.eos_token_id,
                use_cache=True
            )

        # Decode response
        response = tokenizer.decode(outputs[0], skip_special_tokens=True)

        # Extract only the new generated part
        response = response[len(prompt):].strip()

        return response

    except Exception as e:
        print(f"Generation error: {e}")
        return ""

# ========= LLM rules =========
STRICT_RULES = """You are a Text-to-SQL system for SQLite.

STRICT RULES:
1) Output ONLY ONE SQLite SELECT query.
2) Use ONLY the provided tables/columns. NEVER invent names.
3) Use explicit column names (no SELECT *).
4) NEVER use AS aliases - keep original column names.
5) For aggregates use COUNT(*), AVG(col), MIN(col), MAX(col).
6) Return raw SQL only (no explanations, no code fences, no markdown).
7) Do not include any text before or after the SQL query.

Schema:"""

SELECT_WITH_RE = re.compile(r'(?is)\b(select|with)\b.*')

def extract_sql(text: str) -> str:
    if not isinstance(text, str):
        return ""

    # Clean the text
    txt = text.strip()

    # Remove common markdown code fences
    txt = re.sub(r'```sql\s*', '', txt)
    txt = re.sub(r'```\s*', '', txt)

    # Try to find SQL starting with SELECT or WITH
    lines = txt.split('\n')
    sql_lines = []

    for line in lines:
        line = line.strip()
        if line.lower().startswith(('select', 'with')):
            # Found start of SQL
            sql_lines = [line]
            continue
        elif sql_lines:
            # We're in SQL mode
            if line.endswith(';'):
                sql_lines.append(line)
                break
            elif any(keyword in line.lower() for keyword in ['from', 'where', 'group by', 'having', 'order by', 'limit', 'join', 'union', 'and', 'or']):
                sql_lines.append(line)
            elif line == '':
                continue  # Skip empty lines within SQL
            else:
                # Might be end of SQL
                break

    if sql_lines:
        sql = ' '.join(sql_lines)
    else:
        # Fallback to regex
        m = SELECT_WITH_RE.search(txt)
        if m:
            sql = m.group(0).strip()
        else:
            sql = txt

    # Clean up
    if ";" in sql:
        sql = sql.split(";")[0] + ";"

    return sql.strip()

def llm_generate_sql(question: str, db_id: str) -> str:
    prompt = f"""{STRICT_RULES}

{schema_text_for(db_id)}

Question: {question}

SQL:"""

    response = generate_with_qwen(prompt, max_new_tokens=128, temperature=0.1)
    return extract_sql(response)

# ========= ENHANCED REPAIR FUNCTIONS =========
def analyze_question_intent(question: str, db_id: str) -> str:
    """Generate intelligent repair hints based on question analysis"""
    question_lower = question.lower()
    hints = []

    if any(word in question_lower for word in ['how many', 'count', 'number of', 'total']):
        hints.append("- This question asks for COUNT - use COUNT(*) or COUNT(column)")

    if any(word in question_lower for word in ['average', 'mean', 'avg']):
        hints.append("- This question asks for AVERAGE - use AVG(column)")

    if any(word in question_lower for word in ['highest', 'maximum', 'max', 'largest', 'most']):
        hints.append("- This question asks for MAXIMUM - use MAX(column)")

    if any(word in question_lower for word in ['lowest', 'minimum', 'min', 'smallest', 'least']):
        hints.append("- This question asks for MINIMUM - use MIN(column)")

    if any(word in question_lower for word in ['each', 'every', 'per', 'by', 'for each']):
        hints.append("- This question asks for grouping - use GROUP BY")

    if any(word in question_lower for word in ['order', 'sort', 'arrange', 'top', 'bottom']):
        hints.append("- This question asks for ordering - use ORDER BY")

    if any(word in question_lower for word in ['and', 'with', 'having', 'both']):
        hints.append("- This question might require JOINs between tables")

    return "\n".join(hints) if hints else "- Analyze the question structure carefully"

def generate_error_message(gold_result, pred_result):
    """Generate detailed error message for repair system"""
    if isinstance(pred_result, str):
        return pred_result

    error_parts = []

    if gold_result.shape != pred_result.shape:
        error_parts.append(f"Shape mismatch: expected {gold_result.shape}, got {pred_result.shape}")

    gold_cols = [str(c).lower() for c in gold_result.columns]
    pred_cols = [str(c).lower() for c in pred_result.columns]
    if gold_cols != pred_cols:
        error_parts.append(f"Column mismatch: expected {list(gold_result.columns)}, got {list(pred_result.columns)}")

    if gold_result.shape == pred_result.shape and gold_cols == pred_cols:
        try:
            if gold_result.values.tolist() != pred_result.values.tolist():
                error_parts.append("Values don't match expected results")
        except:
            if not gold_result.equals(pred_result):
                error_parts.append("Values don't match expected results")

    return "RESULT ERROR: " + ("; ".join(error_parts) if error_parts else "Unknown difference")

def llm_repair_sql(question: str, db_id: str, bad_sql: str, error_msg: str, gold_result=None, pred_result=None) -> str:
    """Enhanced repair function with intelligent analysis"""
    question_hints = analyze_question_intent(question, db_id)

    prompt = f"""{STRICT_RULES}

{schema_text_for(db_id)}

Question: {question}

REPAIR HINTS:
{question_hints}
- Match column names EXACTLY as they appear in the schema
- Check for typos in column/table names
- Ensure proper SQL syntax and logic

Previous SQL caused error: {error_msg}

Previous SQL: {bad_sql}

Corrected SQL:"""

    response = generate_with_qwen(prompt, max_new_tokens=128, temperature=0.1)
    return extract_sql(response)

def predict_with_repairs(question: str, db_id: str, gold_result=None):
    """Enhanced prediction function with repair mechanism"""
    sql = llm_generate_sql(question, db_id)
    out = run_sql(db_id, sql)
    attempts = 0
    repair_reasons = []
    sql_history = [sql]

    while attempts < MAX_REPAIRS:
        need_repair = False
        error_msg = None

        if isinstance(out, str) and out.startswith("SQL ERROR"):
            need_repair = True
            error_msg = out
            repair_reasons.append("SQL Error")

        elif gold_result is not None and not isinstance(out, str):
            exact, _, _, _ = compare(gold_result, out)
            if not exact:
                need_repair = True
                error_msg = generate_error_message(gold_result, out)
                repair_reasons.append("Wrong Result")

        if not need_repair:
            break

        sql = llm_repair_sql(question, db_id, sql, error_msg, gold_result, out)
        sql_history.append(sql)
        out = run_sql(db_id, sql)
        attempts += 1

    return sql, out, attempts, repair_reasons, sql_history

# ========= Comparison Function =========
def compare(gold, pred):
    """Compare gold and predicted results for exact match"""
    if isinstance(gold, str) or isinstance(pred, str):
        return False, False, False, False

    g, p = gold.reset_index(drop=True), pred.reset_index(drop=True)
    shape_ok = g.shape == p.shape

    gold_cols_lower = [str(c).lower() for c in g.columns]
    pred_cols_lower = [str(c).lower() for c in p.columns]
    cols_ok = shape_ok and gold_cols_lower == pred_cols_lower

    values_match = False
    if shape_ok:
        try:
            values_match = g.values.tolist() == p.values.tolist()
        except:
            values_match = g.equals(p)

    exact = values_match and cols_ok
    return exact, cols_ok, shape_ok, values_match

tick = lambda b: "✅" if b else "❌"

def show_table(df, label):
    if isinstance(df, str):
        print(f"{label}:\n{df}")
    else:
        print(f"{label}:")
        print(df.head(SHOW_ROWS).to_string(index=False))

# ========= MAIN LOOP =========
print(f"🚀 Testing {len(df_dev)} samples with Qwen model")
print("="*70)

orig_exact = imp_exact = 0
orig_mistakes, imp_mistakes = [], []
orig_repair_stats = {"successful": [], "failed": []}
imp_repair_stats = {"successful": [], "failed": []}

for i, row in df_dev.iterrows():
    db_id = row["db_id"]
    q_orig = row["question"]
    q_imp = row.get("question_improved", q_orig)
    gold_sql = row["query"]

    print(f"\n🧪 Row {i+1}/{len(df_dev)} - Database: {db_id}")

    # Golden reference
    print("\n📌 Golden SQL:")
    print(gold_sql)
    gold_out = run_sql(db_id, gold_sql)
    show_table(gold_out, "📌 Golden Table")

    # Original Q with repair system
    print(f"\n❓ Original Q: {q_orig}")
    pred_sql_o, pred_out_o, repair_attempts_o, repair_reasons_o, sql_history_o = predict_with_repairs(q_orig, db_id, gold_out)
    print("🔍 Pred SQL:", pred_sql_o)
    show_table(pred_out_o, "🔍 Pred Table")
    e, c, s, v = compare(gold_out, pred_out_o)
    print(f"🔍 Result → Exact={tick(e)} Cols={tick(c)} Shape={tick(s)} Values={tick(v)}")

    if repair_attempts_o > 0:
        print(f"🔧 Repairs attempted: {repair_attempts_o} ({', '.join(repair_reasons_o)})")
        if e:
            orig_repair_stats["successful"].append(i+1)
            print("✅ Successful repair!")
        else:
            orig_repair_stats["failed"].append(i+1)
            print("❌ Failed repair")

    if not e:
        orig_mistakes.append(i+1)
    orig_exact += e

    # Improved Q with repair system
    print(f"\n✨ Improved Q: {q_imp}")
    pred_sql_i, pred_out_i, repair_attempts_i, repair_reasons_i, sql_history_i = predict_with_repairs(q_imp, db_id, gold_out)
    print("✨ Pred SQL:", pred_sql_i)
    show_table(pred_out_i, "✨ Pred Table")
    e, c, s, v = compare(gold_out, pred_out_i)
    print(f"✨ Result → Exact={tick(e)} Cols={tick(c)} Shape={tick(s)} Values={tick(v)}")

    if repair_attempts_i > 0:
        print(f"🔧 Repairs attempted: {repair_attempts_i} ({', '.join(repair_reasons_i)})")
        if e:
            imp_repair_stats["successful"].append(i+1)
            print("✅ Successful repair!")
        else:
            imp_repair_stats["failed"].append(i+1)
            print("❌ Failed repair")

    if not e:
        imp_mistakes.append(i+1)
    imp_exact += e

    print("="*80)

# ========= FINAL RESULTS =========
N = len(df_dev)
print(f"\n📊 FINAL RESULTS")
print("="*80)

print(f"\n🎯 ACCURACY RESULTS:")
print(f"Original Questions Exact Match: {orig_exact}/{N} = {orig_exact/N*100:.1f}%")
print(f"Improved Questions Exact Match: {imp_exact}/{N} = {imp_exact/N*100:.1f}%")
print(f"Improvement: +{imp_exact - orig_exact} questions (+{(imp_exact - orig_exact)/N*100:.1f}% points)")

print(f"\n🔧 REPAIR STATISTICS:")
orig_total_repairs = len(orig_repair_stats["successful"]) + len(orig_repair_stats["failed"])
imp_total_repairs = len(imp_repair_stats["successful"]) + len(imp_repair_stats["failed"])

print(f"Original Questions:")
print(f"  • Total repair attempts: {orig_total_repairs}")
print(f"  • Successful repairs: {len(orig_repair_stats['successful'])}")
print(f"  • Failed repairs: {len(orig_repair_stats['failed'])}")
if orig_total_repairs > 0:
    print(f"  • Repair success rate: {len(orig_repair_stats['successful'])/orig_total_repairs*100:.1f}%")

print(f"Improved Questions:")
print(f"  • Total repair attempts: {imp_total_repairs}")
print(f"  • Successful repairs: {len(imp_repair_stats['successful'])}")
print(f"  • Failed repairs: {len(imp_repair_stats['failed'])}")
if imp_total_repairs > 0:
    print(f"  • Repair success rate: {len(imp_repair_stats['successful'])/imp_total_repairs*100:.1f}%")

print(f"\n⚠️ REMAINING MISTAKES:")
print("⚠️ ORIGINAL mistakes at rows:", orig_mistakes if orig_mistakes else "None")
print("⚠️ IMPROVED mistakes at rows:", imp_mistakes if imp_mistakes else "None")

print(f"\n✅ SUCCESSFUL REPAIRS:")
if orig_repair_stats["successful"]:
    print(f"Original questions repaired at rows: {orig_repair_stats['successful']}")
if imp_repair_stats["successful"]:
    print(f"Improved questions repaired at rows: {imp_repair_stats['successful']}")

print(f"\n🎉 Evaluation complete with Qwen model!")

# Clean up GPU memory
if torch.cuda.is_available():
    torch.cuda.empty_cache()
````

````python
import os, glob, time, sqlite3, re, pandas as pd
from openai import OpenAI

# ========= CONFIG =========
ROOT = "/content/drive/MyDrive/Dissertition/spider_2_unzipped/spider 2"
DEV_PATH = os.path.join(ROOT, "dev.json")
TABLES_PATH = os.path.join(ROOT, "tables.json")
DB_ROOT = os.path.join(ROOT, "database")
API_KEY_PATH = "/content/drive/MyDrive/Dissertition/spider_2_unzipped/openaikey.txt"

MODEL_NAME = "gpt-4o"
N_SAMPLES = None       # None = full dataset, or set an integer
MAX_REPAIRS = 2      # now allows multiple retries
SLEEP = 0.2
SHOW_ROWS = 5         # rows to print from result

# ========= INIT =========
with open(API_KEY_PATH, "r") as f:
    api_key = f.read().strip()
client = OpenAI(api_key=api_key)

df_dev = pd.read_json(DEV_PATH)
if N_SAMPLES:
    df_dev = df_dev.head(N_SAMPLES)

# ========= DB helpers =========
def db_path_for(db_id: str) -> str:
    paths = [
        os.path.join(DB_ROOT, db_id, f"{db_id}.sqlite"),
        os.path.join(DB_ROOT, f"{db_id}.sqlite")
    ]
    for p in paths:
        if os.path.isfile(p):
            return p
    matches = glob.glob(os.path.join(DB_ROOT, "**", f"{db_id}.sqlite"), recursive=True)
    if matches:
        return sorted(matches)[0]   # stable order
    raise FileNotFoundError(f"Cannot find SQLite for db_id={db_id}")

def is_select_sql(sql: str) -> bool:
    return (sql or "").strip().lower().startswith(("select","with"))

def run_sql(db_id: str, sql: str):
    try:
        if not sql or not isinstance(sql, str):
            return "SQL ERROR: empty SQL"
        if not is_select_sql(sql):
            return "SQL ERROR: only SELECT/CTE queries allowed"
        conn = sqlite3.connect(db_path_for(db_id))
        conn.execute("PRAGMA query_only = ON;")
        df = pd.read_sql_query(sql, conn)
        conn.close()
        return df
    except Exception as e:
        return f"SQL ERROR: {e}"

# ========= LLM rules =========
STRICT_RULES = """You are a Text-to-SQL system for SQLite.

STRICT RULES:
1) Output ONLY ONE SQLite SELECT query.
2) Use explicit column names (no SELECT *).
3) NEVER use AS aliases - keep original column names.
4) For aggregates use COUNT(*), AVG(col), MIN(col), MAX(col).
5) Match golden SQL style exactly.
6) Return raw SQL only (no explanations, no code fences).
"""

SELECT_WITH_RE = re.compile(r'(?is)\b(select|with)\b.*')

def extract_sql(text: str) -> str:
    if not isinstance(text, str): return ""
    txt = text.strip().replace("```sql","").replace("```","").strip()
    m = SELECT_WITH_RE.search(txt)
    if not m: return txt
    sql = m.group(0).strip()
    if ";" in sql:
        sql = sql.split(";",1)[0] + ";"
    return sql

def llm_generate_sql(question: str, db_id: str) -> str:
    prompt = f"""{STRICT_RULES}

Question:
{question}
"""
    r = client.chat.completions.create(
        model=MODEL_NAME,
        messages=[{"role":"user","content":prompt}],
        temperature=0
    )
    return extract_sql(r.choices[0].message.content)

def llm_repair_sql(question: str, db_id: str, bad_sql: str, error_msg: str) -> str:
    prompt = f"""{STRICT_RULES}

Question:
{question}

Previous SQL caused error:
{error_msg}

Previous SQL:
{bad_sql}

Return corrected SQL.
"""
    r = client.chat.completions.create(
        model=MODEL_NAME,
        messages=[{"role":"user","content":prompt}],
        temperature=0
    )
    return extract_sql(r.choices[0].message.content)

def predict_with_repairs(question: str, db_id: str):
    sql = llm_generate_sql(question, db_id)
    out = run_sql(db_id, sql)
    attempts = 0
    while isinstance(out, str) and out.startswith("SQL ERROR") and attempts < MAX_REPAIRS:
        sql = llm_repair_sql(question, db_id, sql, out)
        out = run_sql(db_id, sql)
        attempts += 1
        time.sleep(SLEEP)
    return sql, out

# ========= Comparison =========
def compare(gold, pred):
    if isinstance(gold, str) or isinstance(pred, str):
        return False, False, False, False
    g, p = gold.reset_index(drop=True), pred.reset_index(drop=True)
    shape_ok = g.shape == p.shape
    values_match = shape_ok and g.values.tolist() == p.values.tolist()
    cols_ok = shape_ok and [c.lower() for c in g.columns] == [c.lower() for c in p.columns]
    exact = values_match and cols_ok
    return exact, cols_ok, shape_ok, values_match

tick = lambda b: "✅" if b else "❌"

def show_table(df, label):
    if isinstance(df, str):
        print(f"{label}:\n{df}")
    else:
        print(f"{label}:")
        print(df.head(SHOW_ROWS).to_string(index=False))

# ========= MAIN LOOP =========
print(f"🚀 Testing {len(df_dev)} samples...")
print("="*70)

orig_exact = imp_exact = 0
orig_mistakes, imp_mistakes = [], []

for i, row in df_dev.iterrows():
    db_id = row["db_id"]
    q_orig = row["question"]
    q_imp = row.get("question_improved", q_orig)
    gold_sql = row["query"]

    print(f"\n🧪 Row {i+1}/{len(df_dev)} - Database: {db_id}")

    # Golden reference
    print("\n📌 Golden SQL:")
    print(gold_sql)
    gold_out = run_sql(db_id, gold_sql)
    show_table(gold_out, "📌 Golden Table")

    # Original Q
    print(f"\n❓ Original Q: {q_orig}")
    pred_sql_o, pred_out_o = predict_with_repairs(q_orig, db_id)
    print("🔍 Pred SQL:", pred_sql_o)
    show_table(pred_out_o, "🔍 Pred Table")
    e, c, s, v = compare(gold_out, pred_out_o)
    print(f"🔍 Result → Exact={tick(e)} Cols={tick(c)} Shape={tick(s)} Values={tick(v)}")
    if not e:
        orig_mistakes.append(i+1)   # only row no
    orig_exact += e

    # Improved Q
    print(f"\n✨ Improved Q: {q_imp}")
    pred_sql_i, pred_out_i = predict_with_repairs(q_imp, db_id)
    print("✨ Pred SQL:", pred_sql_i)
    show_table(pred_out_i, "✨ Pred Table")
    e, c, s, v = compare(gold_out, pred_out_i)
    print(f"✨ Result → Exact={tick(e)} Cols={tick(c)} Shape={tick(s)} Values={tick(v)}")
    if not e:
        imp_mistakes.append(i+1)   # only row no
    imp_exact += e

    print("="*80)

# ========= FINAL RESULTS =========
N = len(df_dev)
print("\n📊 FINAL ACCURACY")
print("="*50)
print(f"Original Questions Exact Match: {orig_exact}/{N} = {orig_exact/N*100:.1f}%")
print(f"Improved Questions Exact Match: {imp_exact}/{N} = {imp_exact/N*100:.1f}%")

print("\n⚠️ ORIGINAL mistakes at rows:", orig_mistakes if orig_mistakes else "None")
print("⚠️ IMPROVED mistakes at rows:", imp_mistakes if imp_mistakes else "None")

print("\n🎉 Evaluation complete!")
````
